```python
count = 0
unanswered = 0
while True:
    text = input("You: ").strip().lower()
    count += 1
    if text == "bye":
        break
    elif "hour" in text:
        print("Open 9 to 5")
    else:
        unanswered += 1
print(f"{count} messages, {unanswered} unanswered")
```
